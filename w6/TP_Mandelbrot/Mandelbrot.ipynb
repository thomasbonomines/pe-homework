```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib ipympl
```

```python
def mandelbrot(h, w):
    max = 20
    x = np.linspace(-2, 0.8, w)
    y = np.linspace(-1.4, 1.4, h)
    xx, yy = np.meshgrid(x, y)
    c = xx + 1j * yy
    diverge = np.zeros(c.shape,dtype = int)
    z = np.copy(c)
    for i in range(max):
        mask = np.abs(z) <= 2
        z[mask] = z[mask]**2 + c[mask]
        diverge_a_cette_etape = (np.abs(z) > 2) & (diverge == 0)
        diverge[diverge_a_cette_etape] = i
    egal_0 = (diverge == 0)
    diverge[egal_0] = 20
    plt.figure(figsize=(8, 8))
    plt.imshow(diverge, extent=[-2, 0.8, -1.4, 1.4], cmap='viridis')
    plt.colorbar()
    plt.show()
mandelbrot(1000,1000)
```

```python
def mandelbrot2(xmin, xmax, ymin, ymax, larg, haut, max):
    x = np.linspace(xmin, xmax, larg)
    y = np.linspace(ymin, ymax, haut)
    xx, yy = np.meshgrid(x, y)
    c = xx + 1j * yy
    z = np.copy(c)
    diverge = np.zeros(c.shape,dtype = int)
    for i in range(max):
        mask = np.abs(z) <= 2
        z[mask] = z[mask]**2 + c[mask]
        diverge_a_cette_etape = (np.abs(z) > 2) & (diverge == 0)
        diverge[diverge_a_cette_etape] = i  
    egal_0 = (diverge == 0)
    diverge[egal_0] = max
    plt.figure(figsize=(8, 8))
    plt.imshow(diverge, extent=[xmin, xmax, ymin, ymax], cmap='viridis')
    plt.colorbar()
    plt.show()
```
